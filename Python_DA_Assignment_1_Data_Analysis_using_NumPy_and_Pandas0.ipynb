```python
# Pandas Series: Creating a Series

import pandas as pd

# Create a Pandas Series with marks and custom indices
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

# Display the Pandas Series
print("Student Marks:")
print(marks)

```

```python
# Access the mark of the 1st rank student using integer index position
print("Mark of 1st Rank Student:", marks.iloc[0])

# Retrieve the top 3 ranks using loc
print("Marks of Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Retrieve the mark of the 3rd rank student using iloc
print("Mark of 3rd Rank Student:", marks.iloc[2])

# Filter ranks where marks are greater than 90 using Boolean Mask
print("Ranks with Marks Greater Than 90:")
print(marks[marks > 90])

```

```python
# Modify the mark of the 1st rank student to 100
marks.loc['Rank1'] = 100
print("Marks after modifying 1st rank:")
print(marks)

# Remove the entry corresponding to the last rank
marks = marks.drop('Rank5')
print("\nMarks after removing the last rank:")
print(marks)

# Compute CGPA by dividing each mark by 10
cgpa = marks / 10
print("\nCGPA of Students:")
print(cgpa)

```

```python
# Pandas DataFrame: Creating a DataFrame

import pandas as pd

# Create a DataFrame named transactions
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

# Display the DataFrame
print("Transactions DataFrame:")
print(transactions)
```

```python
# Create the transactions DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

# Display the complete DataFrame
print("Complete Transactions DataFrame:")
print(transactions)

# Display the first 5 rows
print("\nFirst 5 Rows:")
print(transactions.head())

# Display the last 5 rows
print("\nLast 5 Rows:")
print(transactions.tail())

# Display the shape of the DataFrame
print("\nShape of the DataFrame:", transactions.shape)

# Display column names
print("\nColumn Names:")
print(transactions.columns)

# Display data types
print("\nData Types:")
print(transactions.dtypes)

# Display only ProductCategory and Amount columns
print("\nProductCategory and Amount Columns:")
print(transactions[['ProductCategory', 'Amount']])

# Retrieve the last 3 columns using iloc
print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])

# Filter rows where Region is North and Amount is greater than 200
print("\nNorth Region Transactions with Amount Greater Than 200:")
print(transactions[(transactions['Region'] == 'North') &
                   (transactions['Amount'] > 200)])

# Find value counts for ProductCategory
print("\nProduct Category Value Counts:")
print(transactions['ProductCategory'].value_counts())

# Find unique values in Region
print("\nUnique Regions:")
print(transactions['Region'].unique())

# Group by Region and calculate mean Amount
print("\nMean Amount for Each Region:")
print(transactions.groupby('Region')['Amount'].mean())

```

```python
# Modify the Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("After Modifying Amount:")
print(transactions)

# Add a Discount column by calculating 10% of Amount
transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAfter Adding Discount Column:")
print(transactions)

# Remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]
print("\nAfter Removing TransactionID 109:")
print(transactions)

# Delete the Discount column
transactions = transactions.drop(columns=['Discount'])
print("\nAfter Deleting Discount Column:")
print(transactions)

```
